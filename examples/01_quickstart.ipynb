# 01 — Quickstart

`edr-xarray` lets you open an EDR collection as a lazy xarray dataset.
Discover a collection, inspect its structure, then jump to `02_fwi_map.ipynb` for a real plot.


## 1. Discover collections

```python
import httpx

server = "https://edr.example.com"
collections = httpx.get(f"{server}/collections").raise_for_status().json()["collections"]
for c in collections:
    print(c["id"], "-", c.get("title", ""))
```

```python
collection_id = collections[0]["id"]; collection_url = f"{server}/collections/{collection_id}"
```

## 2. Open and inspect

```python
import xarray as xr
import edr_xarray

ds = xr.open_dataset(collection_url, engine="edr")
ds
```

```python
print("dims:     ", dict(ds.sizes))
print("data_vars:", list(ds.data_vars))
```

## 3. Fetch a slice

```python
var = next(iter(ds.data_vars)); ds[var].isel(t=0).values.shape
```
