# 01 - Quickstart

Open an EDR collection as a lazy xarray dataset, inspect what it contains, and
fetch one explicitly selected value.

## 1. Set the server

```python
import httpx
import xarray as xr

import edr_xarray  # registers engine="edr"

server = "https://edr.example.com"
```

## 2. List collections

Run this cell, then choose one collection `id` in the next cell. Use only the `id` value, not the title.

```python
response = httpx.get(f"{server}/collections", timeout=30.0)
response.raise_for_status()
collections = response.json()["collections"]

for collection in collections:
    print("id:   ", collection["id"])
    print("title:", collection.get("title", ""))
    print()
```

## 3. Choose a collection

```python
collection_id = "replace-with-collection-id"  # Use the id value only; do not append the title.
collection_url = f"{server}/collections/{collection_id}"
collection_url
```

## 4. Open lazily

Opening builds the xarray dataset. Data is fetched later, when values are requested.

```python
ds = xr.open_dataset(collection_url, engine="edr")
ds
```

## 5. Inspect structure

Use this output to choose `variable` and `indexer` in the next cell.

```python
print("sizes:")
print(dict(ds.sizes))
print()
print("coords:")
print(list(ds.coords))
print()
print("data variables:")
for name, array in ds.data_vars.items():
    print(name, dict(array.sizes))
```

## 6. Fetch one value

Pick a variable from the list above and provide one integer position for each
of its dimensions. `isel` is zero-based, so a dimension with size `N` accepts
`0` through `N - 1`.


```python
variable = "replace-with-variable-name"
indexer = {
    # Example for a variable with dimensions ("t", "y", "x"):
    # "t": 0,
    # "y": 0,
    # "x": 0,
}

value = ds[variable].isel(indexer).values.item()
value
```

## 7. Close resources

```python
ds.close()
```
