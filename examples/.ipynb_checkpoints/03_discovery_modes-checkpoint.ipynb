# 03 — Discovery modes

When `xr.open_dataset(..., engine="edr")` is called, `edr-xarray` must
figure out the actual coordinate arrays for every axis (longitude,
latitude, vertical levels, time). The library offers **three discovery
strategies**, controlled by the `discovery` keyword:

| Mode | Network cost | Coord resolution | Behavior |
|---|---|---|---|
| `"probe"` *(default)* | +1 cube request | full | fetches a CoverageJSON to read its `domain.axes` |
| `"metadata_only"` | 0 | minimal | uses only the metadata `extent.spatial.bbox` and `extent.temporal.values` |
| `"strict"` | 0 | minimal | like `metadata_only` but raises if `extent.temporal.values` is missing |

This notebook shows the practical difference.

## Setup

We use the standard mock server and the 3D cube payload. The collection
metadata advertises `temporal.values` (three discrete timestamps), so
both `metadata_only` and `strict` will work.

```python
# Mock EDR server — Python stdlib only, no external dependencies.
import json
import threading
from http.server import BaseHTTPRequestHandler, HTTPServer


class _Handler(BaseHTTPRequestHandler):
    """Tiny request handler that serves a routing table of JSON payloads."""

    _routes: dict = {}

    def do_GET(self):  # noqa: N802  (BaseHTTPRequestHandler convention)
        path = self.path.split("?")[0]
        if path in self._routes:
            body = json.dumps(self._routes[path]).encode()
            self.send_response(200)
            self.send_header("Content-Type", "application/json")
            self.send_header("Content-Length", str(len(body)))
            self.end_headers()
            self.wfile.write(body)
        else:
            self.send_response(404)
            self.end_headers()

    def log_message(self, *args, **kwargs):
        # Silence the default stderr access log.
        pass


def _start_server(routes: dict) -> tuple:
    """Start the HTTP server in a daemon thread and return (server, base_url)."""
    _Handler._routes = routes
    srv = HTTPServer(("127.0.0.1", 0), _Handler)
    threading.Thread(target=srv.serve_forever, daemon=True).start()
    return srv, f"http://127.0.0.1:{srv.server_address[1]}"
```

```python
def make_metadata(base_url: str, *, with_vertical: bool = False) -> dict:
    """Build a minimal valid OGC EDR collection metadata document."""
    extent = {
        "spatial": {
            "bbox": [[10.0, 40.0, 11.0, 41.0]],
            "crs": "http://www.opengis.net/def/crs/OGC/1.3/CRS84",
        },
        "temporal": {
            "interval": [[
                "2025-01-01T00:00:00Z",
                "2025-01-03T00:00:00Z",
            ]],
            "values": [
                "2025-01-01T00:00:00Z",
                "2025-01-02T00:00:00Z",
                "2025-01-03T00:00:00Z",
            ],
        },
    }
    if with_vertical:
        extent["vertical"] = {
            "interval": [[1000.0, 500.0]],
            "values": [1000.0, 850.0, 500.0],
            "vrs": "http://www.opengis.net/def/crs/EPSG/0/5714",
        }
    return {
        "id": "demo_collection",
        "title": "Demo Collection",
        "description": "Self-contained demo dataset for the edr-xarray notebook examples.",
        "crs": ["http://www.opengis.net/def/crs/OGC/1.3/CRS84"],
        "output_formats": ["CoverageJSON"],
        "extent": extent,
        "parameter_names": {
            "temperature": {
                "type": "Parameter",
                "unit": {
                    "label": {"en": "Kelvin"},
                    "symbol": {
                        "value": "K",
                        "type": "http://www.opengis.net/def/uom/UCUM/",
                    },
                },
                "observedProperty": {
                    "id": "http://vocab.nerc.ac.uk/standard_name/air_temperature/",
                    "label": {"en": "Air temperature"},
                },
            },
        },
        "data_queries": {
            "cube": {
                "link": {
                    "href": f"{base_url}/collections/demo_collection/cube",
                    "rel": "data",
                    "type": "application/prs.coverage+json",
                    "variables": {
                        "output_formats": ["CoverageJSON"],
                        "default_output_format": "CoverageJSON",
                        "crs_details": [{"crs": "CRS84", "wkt": ""}],
                    },
                }
            }
        },
    }
```

```python
def make_coverage_3d() -> dict:
    """Return a 3D Grid CoverageJSON document with shape (t=3, y=3, x=3)."""
    return {
        "type": "Coverage",
        "domain": {
            "type": "Domain",
            "domainType": "Grid",
            "axes": {
                "x": {"values": [10.0, 10.5, 11.0]},
                "y": {"values": [40.0, 40.5, 41.0]},
                "t": {"values": [
                    "2025-01-01T00:00:00Z",
                    "2025-01-02T00:00:00Z",
                    "2025-01-03T00:00:00Z",
                ]},
            },
            "referencing": [
                {
                    "coordinates": ["x", "y"],
                    "system": {
                        "type": "GeographicCRS",
                        "id": "http://www.opengis.net/def/crs/OGC/1.3/CRS84",
                    },
                },
                {
                    "coordinates": ["t"],
                    "system": {"type": "TemporalRS", "calendar": "Gregorian"},
                },
            ],
        },
        "parameters": {
            "temperature": {
                "type": "Parameter",
                "unit": {
                    "label": {"en": "Kelvin"},
                    "symbol": {
                        "value": "K",
                        "type": "http://www.opengis.net/def/uom/UCUM/",
                    },
                },
                "observedProperty": {
                    "id": "http://vocab.nerc.ac.uk/standard_name/air_temperature/",
                    "label": {"en": "Air temperature"},
                },
            },
        },
        "ranges": {
            "temperature": {
                "type": "NdArray",
                "dataType": "float",
                "axisNames": ["t", "y", "x"],
                "shape": [3, 3, 3],
                "values": [
                    273.15, 274.0, 275.0, 276.0, 277.0, 278.0, 279.0, 280.0, 281.0,
                    272.0, 273.0, 274.0, 275.0, 276.0, 277.0, 278.0, 279.0, 280.0,
                    271.0, 272.0, 273.0, 274.0, 275.0, 276.0, 277.0, 278.0, 279.0,
                ],
            },
        },
    }
```

```python
routes: dict = {}
srv, base_url = _start_server(routes)
routes["/collections/demo_collection"] = make_metadata(base_url)
routes["/collections/demo_collection/cube"] = make_coverage_3d()
collection_url = f"{base_url}/collections/demo_collection"

import xarray as xr
import edr_xarray  # noqa: F401
```

## Mode 1: `probe` (default)

Issues one extra GET to the cube endpoint to retrieve a CoverageJSON,
then reads the actual axis values from `domain.axes`. Result: full
resolution coordinate arrays.

```python
ds_probe = xr.open_dataset(
    collection_url,
    engine="edr",
    discovery="probe",
)
print("probe mode")
print("  dims:", dict(ds_probe.dims))
print("  x:", ds_probe["x"].values)
print("  y:", ds_probe["y"].values)
print("  t:", ds_probe["t"].values)
ds_probe.close()
```

## Mode 2: `metadata_only`

Skips the probe. Coordinate arrays come purely from the metadata
extent: spatial bbox corners (`lon_min, lon_max` for `x`; same for
`y`), and `temporal.values` if present (otherwise the interval
endpoints).

This is the fastest mode (no extra request) but lowest resolution. Use
it when you trust the metadata or when the collection is huge and you
only care about coarse navigation.

```python
ds_meta = xr.open_dataset(
    collection_url,
    engine="edr",
    discovery="metadata_only",
)
print("metadata_only mode")
print("  dims:", dict(ds_meta.dims))
print("  x:", ds_meta["x"].values)  # only 2 corners
print("  y:", ds_meta["y"].values)  # only 2 corners
print("  t:", ds_meta["t"].values)  # all 3 because temporal.values is set
ds_meta.close()
```

## Mode 3: `strict`

Same axis discovery logic as `metadata_only`, but with a runtime check:
if `extent.temporal.values` is missing from the metadata,
`EdrMetadataError` is raised. Useful when your application requires
deterministic, network-free axis discovery.

```python
ds_strict = xr.open_dataset(
    collection_url,
    engine="edr",
    discovery="strict",
)
print("strict mode (success — temporal.values present)")
print("  dims:", dict(ds_strict.dims))
ds_strict.close()
```

### `strict` failure case

Build metadata *without* `temporal.values`, register it, and watch
strict mode raise.

```python
import copy
from edr_xarray import EdrMetadataError

bad_metadata = copy.deepcopy(make_metadata(base_url))
del bad_metadata["extent"]["temporal"]["values"]
routes["/collections/demo_collection"] = bad_metadata

try:
    xr.open_dataset(collection_url, engine="edr", discovery="strict")
except EdrMetadataError as exc:
    print(f"caught EdrMetadataError: {exc}")

# Restore good metadata for the rest of the notebook.
routes["/collections/demo_collection"] = make_metadata(base_url)
```

## Choosing a mode

- **`probe`** — best default. Pays one extra request to get accurate
  coordinate arrays.
- **`metadata_only`** — when you control the server and metadata is
  trustworthy, or when minimizing requests is more important than coord
  precision.
- **`strict`** — same as `metadata_only` but adds a runtime contract
  that the server *must* publish discrete temporal values.

## Cleanup

```python
srv.shutdown()
print("server stopped")
```
